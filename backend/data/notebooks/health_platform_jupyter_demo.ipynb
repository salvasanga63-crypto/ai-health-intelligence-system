## 6. Data loading and Pandas quick examples
Use pandas to load CSV data, inspect the dataset, clean missing values, and group by cluster label.


```python
import pandas as pd
from io import StringIO

csv_data = '''patient_id,first_name,last_name,age,gender,cluster_group,symptoms
1,Jane,Doe,42,Female,women,fever;cough
2,Maria,Johnson,30,Female,pregnant/expectant mothers,abdominal pain;nausea
3,Robert,Kim,67,Male,elders,chest pain;dizziness
4,Amina,Ali,4,Female,children under 5,fever;irritability
5,Ravi,Patel,10,Male,children 5-17,headache;fatigue
'''

df = pd.read_csv(StringIO(csv_data))
df['symptom_count'] = df['symptoms'].str.count(';') + 1
df.head()
```

```python
# Basic cleaning and group operations
df_clean = df.dropna(subset=['cluster_group'])
cluster_summary = df_clean.groupby('cluster_group').agg({'patient_id': 'count', 'symptom_count': 'mean'})
cluster_summary
```

## 7. Visualizations with Matplotlib and interactive plots
Plot sample data inline and switch backends to enable richer output in notebooks.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.plot(df['patient_id'], df['symptom_count'], marker='o')
plt.title('Symptom Count by Patient ID')
plt.xlabel('Patient ID')
plt.ylabel('Symptom Count')
plt.grid(True)
plt.show()
```

```python
# Use an interactive backend if supported by your environment
%matplotlib widget
plt.figure(figsize=(6,3))
plt.bar(df['first_name'], df['symptom_count'], color='skyblue')
plt.title('Symptoms per Patient')
plt.xlabel('Patient')
plt.ylabel('Symptom Count')
plt.show()
```
